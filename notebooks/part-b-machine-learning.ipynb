Our program performs comprehensive analysis of Amazon product reviews. It handles data processing, text analysis, sentiment evaluation, product clustering, and recommendation generation. The system works with Amazon review datasets in CSV format.

The data processing module cleans review text by converting to lowercase, removing special characters, and handling missing values. It normalizes numeric fields including prices and ratings, and processes timestamps for temporal analysis. The code includes specific functions for handling product metadata with nested JSON structures.

For exploratory analysis, the system generates rating distribution histograms and tracks rating trends over time. It identifies products with suspicious rating patterns (high review volume but low scores) through automated detection. Text analysis features include keyword extraction, phrase frequency counting, and word cloud generation to visualize common terms.

-Sentiment analysis

Sentiment analysis combines VADER sentiment scores with normalized star ratings. The system implements multiple classification approaches including Naive Bayes, Random Forest, and BERT models. All models are evaluated using cross-validation with standard metrics (precision, recall, F1-score).


-Advanced Product Analysis & Recommendation Systems

-Product Clustering & Segmentation

The perform_product_clustering() function groups products based on textual and numerical features. It first preprocesses review text by removing stopwords, stemming words, and converting them into TF-IDF vectors. These text features are combined with normalized numerical attributes like price and rating. The system then applies K-Means clustering, automatically determining the optimal number of clusters using the Elbow Method and Silhouette Scores. For visualization, it reduces dimensions using PCA and t-SNE, plotting clusters in 2D space to highlight natural groupings. Finally, it generates a summary of each cluster’s average price and rating, helping identify high-value or underperforming product segments.

-Collaborative Filtering (User & Item-Based)
The system implements two collaborative filtering approaches:

-User-Based CF (user_based_collaborative_filtering()): Recommends products by finding users with similar rating patterns and suggesting items they liked.

-Item-Based CF (item_based_collaborative_filtering()): Recommends products similar to those a user has already rated highly.
Both methods use cosine similarity to measure affinity between users/items and generate predicted ratings. The system evaluates recommendations using Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE) against held-out test data.

-Content-Based Filtering with Word2Vec
The content_based_filtering() function leverages pre-trained Word2Vec embeddings to analyze product descriptions. It converts text into semantic vector representations and computes cosine similarity between products. This allows the system to recommend items with similar descriptions, even if they have no overlapping user ratings.

-Hybrid Recommendations
The hybrid_recommendations() function intelligently merges collaborative and content-based approaches. It normalizes scores from both methods and combines them using adjustable weights, ensuring a balanced recommendation strategy. This hybrid model improves coverage and accuracy, especially for users or products with sparse rating data.

Sentiment Classification & Evaluation
The run_sentiment_classification() function compares multiple models for sentiment prediction:

-Traditional ML models (Naive Bayes, Random Forest) trained on TF-IDF and Word2Vec features.

-BERT-based sentiment analysis via Hugging Face’s pipeline for state-of-the-art accuracy.
Performance is rigorously evaluated using 10-fold cross-validation, reporting precision, recall, F1-score, and accuracy.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import re
import itertools
import os
import nltk
import string
import gensim.downloader as api
from datasets import load_dataset
from collections import Counter
from wordcloud import WordCloud, STOPWORDS
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.metrics import silhouette_score, classification_report, f1_score, accuracy_score, precision_score, recall_score, make_scorer
from sklearn.manifold import TSNE
from sklearn.metrics.pairwise import cosine_similarity
from scipy.sparse import csr_matrix, hstack
from sklearn.model_selection import train_test_split, cross_validate, StratifiedKFold
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.ensemble import RandomForestClassifier
from gensim.models import KeyedVectors
from sklearn.naive_bayes import MultinomialNB
from sklearn.neighbors import KNeighborsClassifier
from transformers import pipeline

def extract_common_keywords_and_phrases(reviews, num_words=20, num_phrases=20):
   
    # Define stopwords to remove common but uninformative words ("the", "and", etc)
    stopwords = set(STOPWORDS)
    words = []
    cleaned_reviews = []
   
    for review in reviews:
        # Convert review to lowercase for uniformity
        review = str(review).lower()
        # Tokenize review into words
        tokens = re.findall(r'\b\w+\b', review)
        # Remove stopwords and short tokens
        filtered_tokens = [word for word in tokens if word not in stopwords and len(word) > 2]
        # Add to word list for word frequency analysis
        words.extend(filtered_tokens)
        # Store cleaned review for phrase extraction later
        cleaned_reviews.append(" ".join(filtered_tokens))

    # Count the most common words
    word_counts = Counter(words)
    common_words = word_counts.most_common(num_words)

    print("\n Most Common Words: \n")
    for word, count in common_words:
        print(f"{word}: {count}")

    # Generate and display a word cloud from the word frequencies
    wordcloud = WordCloud(width=800, height=400, background_color='white', stopwords=stopwords).generate_from_frequencies(word_counts)
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis("off")
    plt.title("Common Words in Low-Rated, High-Reviewed Products")
    plt.show()

    # Extract common bigrams and trigrams (phrases of 2–3 words)
    vectorizer = CountVectorizer(ngram_range=(2,3), stop_words='english', max_features=5000)
    X = vectorizer.fit_transform(cleaned_reviews)
    # Sum total occurrences of each phrase
    sum_words = X.sum(axis=0)
    phrases_freq = [(phrase, int(sum_words[0, idx])) for phrase, idx in vectorizer.vocabulary_.items()]
    # Sort phrases by frequency and keep the top ones
    phrases_freq = sorted(phrases_freq, key=lambda x: x[1], reverse=True)[:num_phrases]

    print("\n Most Common Phrases (Bigrams & Trigrams): \n")
    for phrase, count in phrases_freq:
        print(f"{phrase}: {count}")

def extract_key_attributes_for_top_products(df, top_asins):
    # Iterate through top products (by ASIN) and extract key keywords/phrases from their reviews
    for asin in top_asins.index:
        print(f"\nTop Product ASIN: {asin} - Extracting Key Attributes...\n")
        # Filter reviews for the current ASIN
        product_reviews = df[df["asin"] == asin]["text"].dropna().tolist()
        # Run keyword/phrase analysis on those reviews
        extract_common_keywords_and_phrases(product_reviews)

def clean_data(df):
    # Drop rows with missing review text or rating
    df = df.dropna(subset=["text", "rating"])

    # Normalize prices if a price column exists (remove dollar signs, convert to float)
    if "price" in df.columns:
        # df["price"] = df["price"].replace('[\$,]', '', regex=True).astype(float)
        df.loc[:, "price"] = df["price"].replace('[\$,]', '', regex=True).astype(float)

    # Convert text to lowercase and remove special characters
    df.loc[:, "cleaned_text"] = df["text"].str.lower()
    df.loc[:, "cleaned_text"] = df["cleaned_text"].apply(lambda x: re.sub(r"[^a-zA-Z0-9\s]", "", x))
    # Ensure the rating column contains numeric values
    df.loc[:, "rating"] = pd.to_numeric(df["rating"], errors="coerce")
   
    return df

def convert_category_to_csv(category: str, limit: int = 5, output_dir: str = "converted_reviews"):
    # Construct the config name for the dataset based on the category
    config_name = f"raw_review_{category}"

    # Notify the user that the dataset is being loaded
    print(f"Loading category: {config_name}...")

    # Load the dataset from the Hugging Face hub with streaming
    dataset = load_dataset("McAuley-Lab/Amazon-Reviews-2023", config_name, split="full", streaming=True, trust_remote_code=True)

    # Extract the first `limit` entries from the dataset
    data = list(itertools.islice(dataset, limit))
   
    # Convert the extracted data into a pandas DataFrame
    df = pd.DataFrame(data)

    # Clean the data using the `clean_data` function
    df = clean_data(df)

    # Ensure the output directory exists
    os.makedirs(output_dir, exist_ok=True)
    # Define the output file path
    output_path = os.path.join(output_dir, f"{category}_reviews.csv")
   
    # Save the cleaned DataFrame to a CSV file
    df.to_csv(output_path, index=False)
    print(f"Saved {len(df)} rows to {output_path}")

def convert_metadata_to_csv(category: str, limit: int = 5, output_dir: str = "converted_metadata"):
    import json
    config_name = f"raw_meta_{category}"
    print(f"Loading metadata for category: {config_name}...")

    dataset = load_dataset("McAuley-Lab/Amazon-Reviews-2023", config_name, split="full", trust_remote_code=True)

    data = []
    for example in itertools.islice(dataset, limit):
        for key in [
            "images", "videos", "imageURLs", "tech1", "tech2",
            "similar_items", "also_view", "categories", "related"
        ]:
            if key in example:
                example.pop(key, None)

        if isinstance(example.get("details"), str):
            try:
                example["details"] = json.loads(example["details"])
            except Exception:
                example["details"] = {}

        if isinstance(example.get("authors"), list):
            example["authors"] = [a.get("name", "") for a in example["authors"] if isinstance(a, dict)]

        data.append(example)

    df = pd.DataFrame(data)

    # Flatten 'details' JSON
    if "details" in df.columns:
        details_df = pd.json_normalize(df["details"])
        df = pd.concat([df.drop(columns=["details"]), details_df], axis=1)
   
    # Ensure 'asin' column is present for merge compatibility
    df["asin"] = df["parent_asin"]
   
    # Save CSV
    os.makedirs(output_dir, exist_ok=True)
    output_path = os.path.join(output_dir, f"{category}_metadata.csv")
    df.to_csv(output_path, index=False)
    print(f"Saved {len(df)} metadata rows to {output_path}")

# Ratings and reviews...

def plot_rating_distribution(df, category):
    # Replace infinite values in the 'rating' column with NaN for safe plotting
    df["rating"] = df["rating"].replace([np.inf, -np.inf], np.nan)

    # Create a histogram with kernel density estimation for rating distribution
    plt.figure(figsize=(10, 6))
    sns.histplot(df["rating"], bins=20, kde=True)
    plt.title(f"Rating Distribution for {category}")
    plt.xlabel("Rating")
    plt.ylabel("Frequency")
    plt.show()

def identify_low_ratings_and_high_reviews(df, review_threshold=50, rating_threshold=3):
    # Identify products with high review count (at least `review_threshold` reviews)
    review_counts = df.groupby("asin")["rating"].count()
    high_review_count_products = review_counts[review_counts >= review_threshold].index
   
    # Filter the data to include only those products with high reviews
    high_review_df = df[df["asin"].isin(high_review_count_products)]
   
    # Calculate the average rating for each product
    avg_ratings = high_review_df.groupby("asin")["rating"].mean()
   
    # Identify products with low ratings (below the `rating_threshold`)
    low_rated_products = avg_ratings[avg_ratings <= rating_threshold].index
   
    # Filter the high review products to include only the low-rated ones
    low_rated_with_high_reviews = high_review_df[high_review_df["asin"].isin(low_rated_products)]
   
    # Return the low-rated products with high reviews
    return low_rated_with_high_reviews.groupby("asin").agg({"rating": "mean", "text": "first"}).reset_index()

def plot_top_selling_products(df):
    # Identify top 5 best-selling products by review count
    top_products = df.groupby("asin").size().sort_values(ascending=False).head(5)
   
    if top_products.empty:
        print("\n No products found with reviews in the dataset. \n")
    else:
        print("\n Top 5 best-selling products (sorted by review count): \n")
        # Print the top products without the dtype
        for asin, count in top_products.items():
            print(f"ASIN: {asin}, Review Count: {count}")
    return top_products

def plot_rating_trends(df, category):
    # Assuming we have a timestamp for each review
    df['timestamp'] = pd.to_datetime(df['timestamp'], unit='ms')  # Convert from Unix timestamp to datetime
    df['year_month'] = df['timestamp'].dt.to_period('M')
   
    avg_ratings = df.groupby('year_month')['rating'].mean()
   
    plt.figure(figsize=(10, 6))
    avg_ratings.plot(kind='line')
    plt.title(f"Average Rating Trend for {category}")
    plt.xlabel("Month")
    plt.ylabel("Average Rating")
    plt.grid(True)
    plt.show()

def generate_wordcloud(df):
    # Generate a wordcloud for review text
    text = " ".join(df["cleaned_text"].dropna())
    wordcloud = WordCloud(width=800, height=400, background_color="white").generate(text)
    plt.figure(figsize=(10, 6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.show()

def calculate_sentiment_and_score(df, w1=0.5, w2=0.5):
    analyzer = SentimentIntensityAnalyzer()

    # Safe sentiment extraction
    def get_sentiment(text):
        if isinstance(text, str):
            return analyzer.polarity_scores(text)['compound']
        else:
            return 0.0  # Neutral if text is invalid

    df['text_sentiment'] = df['cleaned_text'].apply(get_sentiment)

    # Normalize ratings (1-5 scale to 0-1)
    df['normalized_rating'] = (df['rating'] - 1) / 4

    # Combine using weighted formula
    df['final_sentiment_score'] = w1 * df['text_sentiment'] + w2 * df['normalized_rating']

    print(df[['text', 'rating', 'text_sentiment', 'normalized_rating', 'final_sentiment_score']])
    return df

#Part 2 

def perform_product_clustering(df, category_name="Unknown", tfidf_max_features=500, max_k=10):
    print(f"\n--- Clustering Products in Category: {category_name} ---")

    # Text preprocessing
    stop_words = set(stopwords.words('english'))
    stemmer = PorterStemmer()

    def preprocess_text(text):
        text = str(text).lower()
        text = re.sub(r"[^a-z\s]", "", text)
        tokens = text.split()
        tokens = [stemmer.stem(word) for word in tokens if word not in stop_words and len(word) > 2]
        return " ".join(tokens)

    df['processed_text'] = df['text'].fillna("").apply(preprocess_text)
    print("Text preprocessing complete.")

    # TF-IDF vectorization
    vectorizer = TfidfVectorizer(max_features=tfidf_max_features)
    text_features = vectorizer.fit_transform(df['processed_text'])
    text_df = pd.DataFrame(text_features.toarray(), columns=vectorizer.get_feature_names_out())
    print(f"TF-IDF vectorization complete.")

    # Scale numeric features
    numeric_features = df[['price', 'rating']].copy()
    numeric_features = numeric_features.fillna(numeric_features.mean())
    scaler = MinMaxScaler()
    scaled_numeric = scaler.fit_transform(numeric_features)
    numeric_df = pd.DataFrame(scaled_numeric, columns=['price_scaled', 'rating_scaled'])
    print(f"Scaled numeric features.")

    # Combine features
    combined_df = pd.concat([text_df.reset_index(drop=True), numeric_df.reset_index(drop=True)], axis=1)
    print(f"Combined feature matrix shape.")

    # Elbow method to determine optimal k
    distortions = []
    silhouette_scores = []
    K = range(2, max_k + 1)

    print("\nComputing optimal number of clusters (elbow & silhouette)...")
    for k in K:
        km = KMeans(n_clusters=k, random_state=42, n_init=10)
        preds = km.fit_predict(combined_df)
        distortions.append(km.inertia_)
        silhouette_scores.append(silhouette_score(combined_df, preds))

    plt.figure(figsize=(12, 5))

    # Plot Inertia, Elbow method
    plt.subplot(1, 2, 1)
    plt.plot(K, distortions, 'bo-')
    plt.xlabel('Number of clusters (k)')
    plt.ylabel('Inertia (Within-cluster sum of squares)')
    plt.title('Elbow Method')

    # Plot Silhouette Score
    plt.subplot(1, 2, 2)
    plt.plot(K, silhouette_scores, 'go-')
    plt.xlabel('Number of clusters (k)')
    plt.ylabel('Silhouette Score')
    plt.title('Silhouette Score vs. k')

    plt.tight_layout()
    plt.show()

    # Choose the best k based on silhouette score
    optimal_k = K[np.argmax(silhouette_scores)]
    print(f"\nOptimal number of clusters based on silhouette score: {optimal_k}")

    # Final KMeans clustering
    kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
    df['cluster'] = kmeans.fit_predict(combined_df)
    print("KMeans clustering complete.")

    # PCA visualization
    pca = PCA(n_components=2)
    reduced_pca = pca.fit_transform(combined_df)

    plt.figure(figsize=(10, 6))
    plt.scatter(reduced_pca[:, 0], reduced_pca[:, 1], c=df['cluster'], cmap='tab10', alpha=0.6)
    plt.title(f"PCA Visualization of Clusters - {category_name}")
    plt.xlabel("PCA Component 1")
    plt.ylabel("PCA Component 2")
    plt.colorbar(label="Cluster")
    plt.grid(True)
    plt.show()

    # t-SNE visualization (optional but often more intuitive for clusters)
    print("Running t-SNE dimensionality reduction:")
    tsne = TSNE(n_components=2, perplexity=30, learning_rate=200, random_state=42)
    reduced_tsne = tsne.fit_transform(combined_df)

    plt.figure(figsize=(10, 6))
    plt.scatter(reduced_tsne[:, 0], reduced_tsne[:, 1], c=df['cluster'], cmap='tab10', alpha=0.6)
    plt.title(f"t-SNE Visualization of Clusters - {category_name}")
    plt.xlabel("t-SNE Component 1")
    plt.ylabel("t-SNE Component 2")
    plt.colorbar(label="Cluster")
    plt.grid(True)
    plt.show()

    # Cluster analysis (mean values)
    print("\nCluster Analysis (average price and rating per cluster):\n")
    cluster_summary = df.groupby('cluster')[['price', 'rating']].mean()
    print(cluster_summary)

    return df

# Collaborative Filtering

def prepare_collaborative_filtering_data(df):
    """Prepare data for collaborative filtering by creating user-item matrix."""
    # Ensure we have the required columns
    if not all(col in df.columns for col in ['user_id', 'asin', 'rating']):
        raise ValueError("DataFrame must contain 'user_id', 'asin', and 'rating' columns")
   
    # Create user-item matrix
    user_item_matrix = df.pivot_table(index='user_id', columns='asin', values='rating', fill_value=0)
    return user_item_matrix

# User based

def user_based_collaborative_filtering(user_item_matrix, user_id, k=5):
    """Implement user-based collaborative filtering to recommend items."""
    # Calculate user-user similarity matrix
    user_similarity = cosine_similarity(user_item_matrix)
    user_similarity_df = pd.DataFrame(user_similarity,
                                     index=user_item_matrix.index,
                                     columns=user_item_matrix.index)
   
    # Get the similarity scores for the given user
    user_similarities = user_similarity_df[user_id]
   
    # Remove the user's own similarity score
    user_similarities = user_similarities.drop(user_id)
   
    # Get top k similar users
    similar_users = user_similarities.nlargest(k)
   
    # Get items rated by similar users but not by our target user
    target_user_items = user_item_matrix.loc[user_id]
    similar_users_items = user_item_matrix.loc[similar_users.index]
   
    # Calculate weighted average of ratings from similar users
    recommendations = {}
    for item in similar_users_items.columns:
        if target_user_items[item] == 0:  # Item not rated by target user
            # Get ratings from similar users who rated this item
            item_ratings = similar_users_items[item]
            rated_users = item_ratings[item_ratings > 0].index
           
            if len(rated_users) > 0:
                weighted_sum = sum(user_similarities[u] * similar_users_items.loc[u, item] for u in rated_users)
                similarity_sum = sum(user_similarities[u] for u in rated_users)
                if similarity_sum > 0:
                    predicted_rating = weighted_sum / similarity_sum
                    recommendations[item] = predicted_rating
   
    # Sort recommendations by predicted rating
    sorted_recommendations = sorted(recommendations.items(), key=lambda x: x[1], reverse=True)
   
    return sorted_recommendations

# Item based

def item_based_collaborative_filtering(user_item_matrix, user_id, k=5):
    """Implement item-based collaborative filtering to recommend items."""
    # Transpose the matrix to get item-user matrix
    item_user_matrix = user_item_matrix.T
   
    # Calculate item-item similarity matrix
    item_similarity = cosine_similarity(item_user_matrix)
    item_similarity_df = pd.DataFrame(item_similarity,
                                     index=item_user_matrix.index,
                                     columns=item_user_matrix.index)
   
    # Get items the user has already rated
    user_ratings = user_item_matrix.loc[user_id]
    rated_items = user_ratings[user_ratings > 0].index
   
    # For each unrated item, calculate predicted rating
    recommendations = {}
    for item in item_user_matrix.index:
        if user_ratings[item] == 0:  # Item not rated by user
            # Find most similar items that the user has rated
            similar_items = item_similarity_df[item].drop(item)
            similar_items_rated = similar_items[similar_items.index.isin(rated_items)]
           
            if len(similar_items_rated) > 0:
                top_similar_items = similar_items_rated.nlargest(k)
                weighted_sum = sum(top_similar_items[i] * user_ratings[i] for i in top_similar_items.index)
                similarity_sum = sum(top_similar_items[i] for i in top_similar_items.index)
                if similarity_sum > 0:
                    predicted_rating = weighted_sum / similarity_sum
                    recommendations[item] = predicted_rating
   
    # Sort recommendations by predicted rating
    sorted_recommendations = sorted(recommendations.items(), key=lambda x: x[1], reverse=True)
   
    return sorted_recommendations

def evaluate_recommendations(recommendations, test_ratings, top_k=5):
    """Evaluate recommendation quality by comparing with actual test ratings."""
    # Convert recommendations to DataFrame
    rec_df = pd.DataFrame(recommendations, columns=['asin', 'predicted_rating'])
   
    # Merge with test ratings
    evaluation_df = pd.merge(rec_df.head(top_k), test_ratings, on='asin', how='left')
   
    # Calculate evaluation metrics
    if not evaluation_df.empty and 'rating' in evaluation_df.columns:
        mae = (evaluation_df['predicted_rating'] - evaluation_df['rating']).abs().mean()
        rmse = ((evaluation_df['predicted_rating'] - evaluation_df['rating'])**2).mean()**0.5
        print(f"Evaluation Metrics (Top {top_k} recommendations):")
        print(f"MAE: {mae:.3f}")
        print(f"RMSE: {rmse:.3f}")
    else:
        print("No overlapping items between recommendations and test set for evaluation.")
   
    return evaluation_df

# Content-Based filtering

# Load pre-trained Word2Vec model
word2vec_model = api.load("word2vec-google-news-300")

# Helper: Clean and tokenize text
def process_text(text):
    if pd.isnull(text):
        return []
    tokens = word_tokenize(text.lower())
    tokens = [t for t in tokens if t.isalpha() and t not in stopwords.words('english')]
    return tokens

# Convert a product description into an averaged Word2Vec vector
def get_description_vector(description, model):
    words = process_text(description)
    valid_vectors = [model[word] for word in words if word in model]
    if valid_vectors:
        return np.mean(valid_vectors, axis=0)
    else:
        return np.zeros(model.vector_size)

# Generate content-based recommendations
def content_based_filtering(df, product_id, top_k=5):
    """Recommend similar products based on description content."""
    if 'description' not in df.columns:
        raise ValueError("DataFrame must contain a 'description' column.")

    print("Computing description vectors...")
    df = df.dropna(subset=['description']).drop_duplicates(subset='asin').reset_index(drop=True)
   
    # Store asin to index mapping
    asin_to_index = {asin: idx for idx, asin in enumerate(df['asin'])}
   
    if product_id not in asin_to_index:
        raise ValueError("Invalid product ID provided.")

    # Compute description vectors
    df['desc_vector'] = df['description'].apply(lambda x: get_description_vector(x, word2vec_model))
    matrix = np.stack(df['desc_vector'].values)
   
    # Compute cosine similarity
    similarity_matrix = cosine_similarity(matrix)

    # Get index of the target product
    target_index = asin_to_index[product_id]
    similarity_scores = similarity_matrix[target_index]

    # Get top K similar products
    top_indices = np.argsort(similarity_scores)[::-1][1:top_k+1]
    similar_products = df.iloc[top_indices][['asin', 'title', 'description']].copy()
    similar_products['similarity_score'] = similarity_scores[top_indices]

    return similar_products

# Hybrid

def hybrid_recommendations(cf_recs, cbf_recs, cf_weight=0.7, cbf_weight=0.3, top_k=5):
    # Convert CF to DataFrame
    cf_df = pd.DataFrame(cf_recs, columns=['asin', 'cf_score'])
   
    # Rename cbf_score
    cbf_df = cbf_recs[['asin', 'similarity_score']].rename(columns={'similarity_score': 'cbf_score'})

    # Normalize scores to [0, 1] for fair weighting
    scaler = MinMaxScaler()
    if not cf_df.empty:
        cf_df['cf_score'] = scaler.fit_transform(cf_df[['cf_score']])
    if not cbf_df.empty:
        cbf_df['cbf_score'] = scaler.fit_transform(cbf_df[['cbf_score']])

    # Merge on 'asin'
    hybrid_df = pd.merge(cf_df, cbf_df, on='asin', how='inner')  # only combine where both CF and CBF exist

    if hybrid_df.empty:
        print("No overlapping recommendations between CF and CBF.")
        return pd.DataFrame()

    # Compute hybrid score
    hybrid_df['hybrid_score'] = (cf_weight * hybrid_df['cf_score']) + (cbf_weight * hybrid_df['cbf_score'])

    # Return top-k
    return hybrid_df.sort_values(by='hybrid_score', ascending=False).head(top_k)

def recommendation_system_demo(df, category_name):
    print(f"\n--- Recommendation System Demo for {category_name} ---")
   
    # Initialize placeholders for different types of recommendations
    user_based_rec = []
    item_based_rec = []
    cbf_recs = pd.DataFrame()
    demo_user = None
    sample_product_id = None

    # Ensure the necessary columns exist for collaborative filtering
    if all(col in df.columns for col in ['user_id', 'asin', 'rating']):
        # Count number of interactions per user and per item
        user_counts = df['user_id'].value_counts()
        item_counts = df['asin'].value_counts()
        
        # Filter to keep only active users and popular items
        active_users = user_counts[user_counts >= 3].index
        active_items = item_counts[item_counts >= 5].index
        filtered_df = df[df['user_id'].isin(active_users) & df['asin'].isin(active_items)]

        # Proceed only if there's enough data after filtering
        if len(filtered_df) >= 100:
            # Split the data into training and testing sets
            train_df, test_df = train_test_split(filtered_df, test_size=0.2, random_state=42)
            
            # Prepare user-item matrix for collaborative filtering
            train_matrix = prepare_collaborative_filtering_data(train_df)

            # Randomly choose a user to demonstrate recommendations
            demo_user = np.random.choice(train_matrix.index)
            print(f"\nDemonstrating recommendations for user: {demo_user}")

            # ----- User-Based Collaborative Filtering -----
            print("\nUser-Based Collaborative Filtering Recommendations:")
            user_based_rec = user_based_collaborative_filtering(train_matrix, demo_user, k=5)
            for i, (item, rating) in enumerate(user_based_rec, 1):
                print(f"{i}. Item: {item}, Predicted Rating: {rating:.2f}")

            # Evaluate recommendations using test set
            user_test_ratings = test_df[test_df['user_id'] == demo_user][['asin', 'rating']]
            evaluate_recommendations(user_based_rec, user_test_ratings)

            # ----- Item-Based Collaborative Filtering -----
            print("\nItem-Based Collaborative Filtering Recommendations:")
            item_based_rec = item_based_collaborative_filtering(train_matrix, demo_user, k=5)
            for i, (item, rating) in enumerate(item_based_rec, 1):
                print(f"{i}. Item: {item}, Predicted Rating: {rating:.2f}")

            # Evaluate item-based CF recommendations
            evaluate_recommendations(item_based_rec, user_test_ratings)
        else:
            print("Not enough data for collaborative filtering after filtering.")
    else:
        print("Skipping collaborative filtering - missing required columns (user_id, asin, rating)")

    # --------- Content-Based Filtering Demo ---------
    print("\nContent-Based Filtering Recommendations:")

    # Check if product descriptions exist for CBF
    if 'description' not in df.columns:
        print("Skipping content-based filtering - 'description' column not found.")
        return

    # Remove entries with missing descriptions and duplicates
    content_df = df.dropna(subset=['description']).drop_duplicates(subset='asin')
    
    # Ensure we have enough products with descriptions
    if len(content_df) < 10:
        print("Not enough products with valid descriptions for content-based filtering.")
        return

    # Randomly pick a product to base recommendations on
    sample_product_id = np.random.choice(content_df['asin'].values)
    print(f"Sample product for content-based filtering: {sample_product_id}")

    try:
        # Generate content-based recommendations
        cbf_recs = content_based_filtering(content_df, sample_product_id, top_k=15)
        for i, row in enumerate(cbf_recs.itertuples(), 1):
            print(f"{i}. ASIN: {row.asin}, Title: {row.title}, Similarity Score: {row.similarity_score:.3f}")
    except Exception as e:
        print(f"Error during content-based filtering: {e}")
        cbf_recs = pd.DataFrame()

    # --------- Hybrid Recommendations ---------
    print("\nHybrid Recommendations:")

    # Prefer user-based recommendations, fall back to item-based if not available
    cf_recs = user_based_rec if user_based_rec else item_based_rec

    # Generate hybrid recommendations only if both CF and CBF outputs exist
    if cf_recs and not cbf_recs.empty:
        try:
            # Combine CF and CBF scores into a hybrid score
            hybrid_df = hybrid_recommendations(cf_recs, cbf_recs, cf_weight=0.7, cbf_weight=0.3, top_k=5)
            
            # Print the top hybrid recommendations
            if not hybrid_df.empty:
                for i, row in enumerate(hybrid_df.itertuples(), 1):
                    print(f"{i}. ASIN: {row.asin}, Hybrid Score: {row.hybrid_score:.3f}")
            else:
                print("No hybrid recommendations could be generated.")
        except Exception as e:
            print(f"Error during hybrid recommendation computation: {e}")
    else:
        print("Skipping hybrid recommendations - incomplete CF or CBF results.")


def evaluate_model_with_cv(classifier, X, y):
    # Define the metrics we want to evaluate: precision, recall, f1-score, and accuracy
    scoring = ['precision_macro', 'recall_macro', 'f1_macro', 'accuracy']

    # Set up stratified 10-fold cross-validation to preserve label proportions in each fold
    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

    # Perform cross-validation with custom scoring metrics
    scores = cross_validate(
        classifier,       # The classifier to evaluate
        X, y,             # Features and labels
        cv=cv,            # Cross-validation strategy
        scoring={         # Scoring metrics with specific settings
            'precision_macro': make_scorer(precision_score, average='macro', zero_division=0),
            'recall_macro': make_scorer(recall_score, average='macro', zero_division=0),
            'f1_macro': make_scorer(f1_score, average='macro'),
            'accuracy': make_scorer(accuracy_score)
        }
    )

    # Return the average of each score across all folds, rounded to 4 decimal places
    return {
        "Precision": round(scores["test_precision_macro"].mean(), 4),
        "Recall": round(scores["test_recall_macro"].mean(), 4),
        "F1-Score": round(scores["test_f1_macro"].mean(), 4),
        "Accuracy": round(scores["test_accuracy"].mean(), 4)
    }

def run_sentiment_classification(df):
    cv_table = []   # Store cross-validation results
    test_table = [] # Store final test results

    # Map numeric ratings to sentiment labels
    def label_sentiment(rating):
        if rating >= 4:
            return "positive"
        elif rating == 3:
            return "neutral"
        else:
            return "negative"

    # Apply sentiment labeling to the dataset
    df["sentiment"] = df["rating"].apply(label_sentiment)

    # Encode sentiment labels into integers
    label_encoder = LabelEncoder()
    y = label_encoder.fit_transform(df["sentiment"])

    # Create TF-IDF features from the processed text
    tfidf_vectorizer = TfidfVectorizer(max_features=5000)
    X_tfidf = tfidf_vectorizer.fit_transform(df["processed_text"])

    # Convert each text into a Word2Vec vector by averaging word vectors
    def get_word_vector(text, model):
        tokens = process_text(text)
        vectors = [model[word] for word in tokens if word in model]
        return np.mean(vectors, axis=0) if vectors else np.zeros(model.vector_size)

    # Apply Word2Vec vectorization across the dataset
    X_word2vec = np.vstack(df["processed_text"].apply(lambda x: get_word_vector(x, word2vec_model)))

    # Split data into training and testing sets for both feature types
    X_train_tfidf, X_test_tfidf, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.2, random_state=42)
    X_train_w2v, X_test_w2v, _, _ = train_test_split(X_word2vec, y, test_size=0.2, random_state=42)

    # Define classifiers to evaluate
    classifiers = {
        "Naive Bayes": MultinomialNB(),
        "KNN": KNeighborsClassifier(n_neighbors=5),
        "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42)
    }

    # Define feature sets to test with classifiers
    feature_sets = {
        "TF-IDF": (X_train_tfidf, X_test_tfidf),
        "Word2Vec": (X_train_w2v, X_test_w2v)
    }

    print("\n Running classic ML models with 10-fold CV and final test evaluation:\n")

    # Loop over classifiers and feature sets
    for model_name, clf in classifiers.items():
        for feat_name, (X_train, X_test) in feature_sets.items():
            # Skip incompatible combination: Naive Bayes with Word2Vec
            if model_name == "Naive Bayes" and feat_name == "Word2Vec":
                continue

            # Evaluate model using 10-fold cross-validation
            cv_results = evaluate_model_with_cv(clf, X_train, y_train)

            # Train model and evaluate on held-out test set
            clf.fit(X_train, y_train)
            y_pred = clf.predict(X_test)
            test_precision = precision_score(y_test, y_pred, average="macro", zero_division=0)
            test_recall = recall_score(y_test, y_pred, average="macro", zero_division=0)
            test_f1 = f1_score(y_test, y_pred, average="macro")
            test_acc = accuracy_score(y_test, y_pred)

            # Store cross-validation and test results
            cv_table.append({
                "Feature Set": feat_name,
                "Model": model_name,
                **cv_results
            })

            test_table.append({
                "Feature Set": feat_name,
                "Model": model_name,
                "Precision": round(test_precision, 4),
                "Recall": round(test_recall, 4),
                "F1-Score": round(test_f1, 4),
                "Accuracy": round(test_acc, 4)
            })

            print(f"{model_name} ({feat_name}) -> Test F1: {test_f1:.4f}, CV F1: {cv_results['F1-Score']}")

    # BERT sentiment analysis using a HuggingFace pipeline
    try:
        print("\n Running BERT model (nlptown/bert-base-multilingual-uncased-sentiment)...")

        # Load sentiment-analysis pipeline
        hf_pipeline = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")

        # Use a sample of texts for prediction
        test_texts = df["text"].iloc[:200].tolist()
        preds = hf_pipeline(test_texts, truncation=True)

        # Map BERT prediction labels to "positive", "neutral", "negative"
        mapped_preds = []
        for p in preds:
            label = p["label"]
            if "1" in label or "2" in label:
                mapped_preds.append("negative")
            elif "3" in label:
                mapped_preds.append("neutral")
            else:
                mapped_preds.append("positive")

        # Encode predictions and compute metrics
        mapped_preds_encoded = label_encoder.transform(mapped_preds[:len(mapped_preds)])
        true_labels_encoded = y[:len(mapped_preds)]
        f1 = f1_score(true_labels_encoded, mapped_preds_encoded, average="macro")
        acc = accuracy_score(true_labels_encoded, mapped_preds_encoded)
        prec = precision_score(true_labels_encoded, mapped_preds_encoded, average="macro")
        rec = recall_score(true_labels_encoded, mapped_preds_encoded, average="macro")

        # Store BERT evaluation results
        test_table.append({
            "Feature Set": "Raw Text",
            "Model": "BERT (HF Pipeline)",
            "Precision": round(prec, 4),
            "Recall": round(rec, 4),
            "F1-Score": round(f1, 4),
            "Accuracy": round(acc, 4)
        })

        print(f"BERT Test F1-score: {f1:.4f}")

    except Exception as e:
        # Handle failure gracefully
        print(f"BERT pipeline failed: {e}")

    # Convert results to DataFrames for display or export
    cv_df = pd.DataFrame(cv_table)
    test_df = pd.DataFrame(test_table)

    print("\n 10-Fold Cross-Validation Results:")
    print(cv_df)

    print("\n Final Test Set Evaluation:")
    print(test_df)

    return cv_df, test_df

def main():
    # Categories of Amazon product reviews to process
    categories = ["All_Beauty", "Digital_Music", "Gift_Cards", "Health_and_Personal_Care", "Magazine_Subscriptions"]
    rows_per_category = 10000   # Number of rows (reviews) to process per category

    category_avg_ratings = {}  # Store average ratings per category

    for cat in categories:
        try:
            print(f"\nProcessing category: {cat}")
            
            # Convert raw JSON review and metadata files to CSV format
            convert_category_to_csv(cat, limit=rows_per_category)
            convert_metadata_to_csv(cat, limit=rows_per_category)

            # Load converted review and metadata files
            df = pd.read_csv(f"converted_reviews/{cat}_reviews.csv")
            meta_df = pd.read_csv(f"converted_metadata/{cat}_metadata.csv", low_memory=False)

            # Merge price data from metadata into the review DataFrame using 'asin' (product ID)
            if 'asin' in df.columns and 'asin' in meta_df.columns:
                df = df.merge(meta_df[['asin', 'price']], on='asin', how='left')
            else:
                raise KeyError("'asin' column missing from one of the datasets.")

            # Merge product descriptions if available
            if 'description' in meta_df.columns:
                df = df.merge(meta_df[['asin', 'description']], on='asin', how='left')
            else:
                print("Warning: 'description' column not found in metadata.")

            # Perform product clustering based on features 
            df = perform_product_clustering(df, category_name=cat)

            print("\n Rating Distribution: \n")
            # Plot distribution of ratings 
            plot_rating_distribution(df, cat)

            # Identify products with low ratings but high number of reviews
            low_rated_df = identify_low_ratings_and_high_reviews(df)
            print(f"Low-rated products with high reviews: {low_rated_df}")

            # Extract commonly mentioned keywords/phrases from negative reviews
            if not low_rated_df.empty:
                reviews = low_rated_df['text'].dropna().tolist()
                extract_common_keywords_and_phrases(reviews)
            else:
                print("No low-rated, high-review products found in this category.")

            # Identify and visualize top-selling products
            top_products = plot_top_selling_products(df)

            # Extract key attributes for those top-selling products
            extract_key_attributes_for_top_products(df, top_products)

            print("\n Average Rating Rate: \n")
            # Plot how average rating changes over time 
            plot_rating_trends(df, cat)

            print("\n Wordclouds: \n")
            # Generate word clouds for positive/negative reviews
            generate_wordcloud(df)

            print("\n Sentiment Score: \n")
            # Compute sentiment score using weighted combination of metrics
            df = calculate_sentiment_and_score(df, w1=0.6, w2=0.4)

            print("\n Recommendation System Demo: \n")
            # Show personalized or content-based product recommendations
            recommendation_system_demo(df, cat)

            # Preprocess review text for classification 
            df["processed_text"] = df["text"].fillna("").apply(process_text).apply(lambda x: " ".join(x))

            print("\n Running Sentiment Classification: \n")
            # Train and evaluate classifiers to predict sentiment (positive/neutral/negative)
            run_sentiment_classification(df)

            # Calculate and store the average rating for this category
            avg_rating = df['rating'].mean()
            category_avg_ratings[cat] = avg_rating

        except Exception as e:
            # Catch any errors and continue with next category
            print(f"Failed to process {cat}: {e}")

    # After all categories are processed
    if category_avg_ratings:
        print("\n Average Ratings by Category:")
        for cat, avg in category_avg_ratings.items():
            print(f"{cat}: {avg:.2f}")

        # Visualize average ratings per category
        plt.figure(figsize=(8, 5))
        sns.barplot(x=list(category_avg_ratings.keys()), y=list(category_avg_ratings.values()))
        plt.title("Average Rating by Category")
        plt.ylabel("Average Rating")
        plt.ylim(0, 5)
        plt.xticks(rotation=30)
        plt.grid(axis='y')
        plt.tight_layout()
        plt.show()

        # Compute overall average rating across all categories
        overall_mean = np.mean(list(category_avg_ratings.values()))
        print(f"\n Overall Average Rating: {overall_mean:.2f}")

        # Show categories with above-average ratings
        print("\n Categories with higher-than-average ratings:")
        for cat, avg in category_avg_ratings.items():
            if avg > overall_mean:
                print(f"{cat}: {avg:.2f}")

        # Show categories with below-average ratings
        print("\n Categories with lower-than-average ratings:")
        for cat, avg in category_avg_ratings.items():
            if avg < overall_mean:
                print(f"{cat}: {avg:.2f}")

if __name__ == "__main__":
    main()
```
